```python
import json

# Load both datasets
with open(r"D:\student1402\negar\final_research\data\lotus_seed_dataset.json") as f:
    lotus_data = json.load(f)

with open(r"D:\student1402\negar\final_research\data\curated_test_set.json") as f:
    test_data = json.load(f)

# Get PMIDs from lotus_seed_dataset (convert to string for consistent comparison)
lotus_pmids = set(str(item["pubmed_id"]) for item in lotus_data)

# Get PMIDs from curated_test_set (keys are already PMIDs as strings)
test_pmids = set(test_data.keys())

# Check overlap
overlap = lotus_pmids & test_pmids

print(f"Total unique PMIDs in lotus_seed_dataset: {len(lotus_pmids)}")
print(f"Total PMIDs in curated_test_set: {len(test_pmids)}")
print(f"Number of overlapping PMIDs: {len(overlap)}")
print(f"Overlapping PMIDs: {overlap}")
```

```python
import json


# Load both datasets
with open(r"D:\student1402\negar\final_research\data\lotus_seed_dataset.json") as f:
    lotus_data = json.load(f)

with open(r"D:\student1402\negar\final_research\data\curated_test_set.json") as f:
    test_data = json.load(f)

test_pmids = set(test_data.keys())

# Samples whose PMID overlaps with the test set
overlap_samples = [item for item in lotus_data if str(item["pubmed_id"]) in test_pmids]

# Samples NOT in the test set -> new clean raw dataset
clean_samples = [item for item in lotus_data if str(item["pubmed_id"]) not in test_pmids]

print(f"Total samples: {len(lotus_data)}")
print(f"Overlap samples: {len(overlap_samples)}")
print(f"Clean samples: {len(clean_samples)}")

with open("lotus_overlap_samples.json", "w") as f:
    json.dump(overlap_samples, f, ensure_ascii=False, indent=2)

with open("lotus_seed_dataset_clean.json", "w") as f:
    json.dump(clean_samples, f, ensure_ascii=False, indent=2)
```

```python
import json

# Load datasets
with open(r"D:\student1402\negar\final_research\data\lotus_seed_dataset.json") as f:
    lotus_data = json.load(f)

with open(r"D:\student1402\negar\final_research\data\curated_test_set.json") as f:
    test_data = json.load(f)

test_pmids = set(test_data.keys())

# Keep only samples NOT in the test set
clean_samples = [item for item in lotus_data if str(item["pubmed_id"]) not in test_pmids]

print(f"Total samples: {len(lotus_data)}")
print(f"Removed (overlap): {len(lotus_data) - len(clean_samples)}")
print(f"Remaining: {len(clean_samples)}")

# Save new clean dataset
with open(r"D:\student1402\negar\final_research\data\lotus_seed_dataset_clean.json", "w", encoding="utf-8") as f:
    json.dump(clean_samples, f, ensure_ascii=False, indent=2)

print("Saved to lotus_seed_dataset_clean.json")
```
